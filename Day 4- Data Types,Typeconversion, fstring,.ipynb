## Data Types: 
### Int, float, complex, bool, str, list, tuple, range, set, dic,etc

```python
## Int example
age =25
count=10
print(age)
print("age")
print(type(age))
```

```python
## float example
number=25.0
print(number)
print(type(number))
```

```python
## Complex contains imaginary component
number =2+1j
print(number)
print(type(number))
```

```python
## Boolean python ha stwo boolean values Truth or False.
is_active= True
is_complete= False
print(type(is_active))
```

```python
# string
a="fde"
b="45"
c="-455"
print(a)
print(b)
print(type(c))
```

```python
## List is an order and mutable collection and allow duplicate values[].
number = ['a','b','c','d','e']
print(number)
print(type(number))

```

```python
## Tuple: Itis an order collection and imutable()
number=(910,20,30)
print(number)
print(type(number))
```

```python
#Set : mutable collection of unique element.{}
number ={10, 20, 30,30}
print(number)
print(type(number))
```

```python
# Dictionary: Stores key value pairs.{}
student ={"name":"ravi",
          "age":20,
          "Gender":"male"
         }
print(student)
print(type(student))
```

```python
## None Type:Absence of the value
result= None
print(result)
print(type(result))
```

```python
#Range:
number= range(5)
print(number)
print(list(number))
print(type(number))
```

## Conversion Type process
## Coverting a value from one data type to another data type.


```python
a=2
b=3.5
c=a+b
print(c)
print(type(c))
```

```python
##string to int
age ="25"
age=int(age)
print(age)
print(type(age))
```

```python
#Float to Int conversion
a=2
b=3.5
c=int(a+b)
print(c)
print(type(c))
```

```python
#string to  list
age ="25"
age=list(age)
print(age)
print(type(age))
```

```python
#string to tuple
age ="25"
age=tuple(age)
print(age)
print(type(age))
```

```python
#string to set
age ="25"
age=set(age)
print(age)
print(type(age))
```

```python
#string to float
age ="25"
age=float(age)
print(age)
print(type(age))
```

```python
#string to float
age ="hello"
age=float(age)
print(age)
print(type(age))
```
