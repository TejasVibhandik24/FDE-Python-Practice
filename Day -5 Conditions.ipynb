```python
marks =75
if marks>=90:
    print("grade A")
elif marks>=75:
    print("grade B")
elif marks>=60:
    print("grade C")
else:
    print("fail")
## Execute only one matching block only.
```

## Nested Conditions:
## Condition inside condition

```python
## Nested Condition
marks = 80
lab = 180
if marks>=60:
    print("pass")
    if lab>=90:
        print("Eligible for 1st grade")
    else:
        print("Not eligible for 1st grade")
else :
    print("fail")
```
