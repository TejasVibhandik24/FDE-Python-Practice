### String Immutability
### String can not be modified once we declared

```python
# String immutable means you cannot modify the existing string after it is created.
word ="PYTHON"
word[0]="J"
print(word)
```

```python
word ="PYTHON"
new_word ="J"+ word[1:]
print("old:",word)
print("new:",new_word)
```

```python
## Join
word = ["python", "is","easy"]
sentence = " ".join(word)
print(word)
print(sentence)
```

```python
date="2026-11-07"
parts=date.split("-")
print(parts)
```

### Mutable : add, update, remove, insert, extend, sort.

```python
# update
fruit = ["apple","banana","orange"]
print(fruit[0])
```

```python
fruit[0]="mango"
print(fruit)
## here, python replaces apple with mango. So, it is mutable.
```

```python
fruit = ("apple","banana","orange")
print(fruit[0])
```

```python
## Append: By default adds a new elememt at the end of the list.
fruit = ["apple","banana","orange"]
fruit.append("mango") ##function method used parenthesis()
print(fruit)
```

```python
## append with multiple elements example (adds at the end of the list.)
fruit = ["apple","banana","orange"]
fruit.append(["mango","cherry"])  ## python adds the entire list consider as one single element
print(fruit)
```

```python
## Extend : Adds as a individual elements also add as a last element
fruit = ["apple","banana","orange"]
fruit.extend(["mango","cherry"])  
print(fruit)
```

```python
## Insert method : Adds the element into specific position
fruit = ["apple","banana","orange"]
fruit.insert(1,"mango")
print(fruit)
```

```python
fruit = ["apple","banana","orange"]
fruit.insert(3,"grape")
print(fruit)
```

```python
## Remove() Method:remove element by value
## Remove method not supports element by index.
fruit = ["apple","banana","orange"]
fruit.remove("banana")
print(fruit)
```

```python
## Delete method: We can use del method to delete specific element by calling index.
## Del mthod not support by elememt value.
fruit = ["apple","banana","orange"]
del fruit[1]
print(fruit)
```

```python
## Pop Method: Remove last element
fruit = ["apple","banana","orange"]
fruit.pop()
print(fruit)
```

## Pop Method: Remove using index specific elememt we can remove 
fruit = ["apple","banana","orange"]
fruit.pop(1)
print(fruit)

```python
## Sort : Arrange the elemet in ascending order.
## Pop Method: Remove last element
fruit = ["banana","apple","orange"]
fruit.sort()
print(fruit)
```

```python
print(fruit[0:2]) ## slicing 
```

## Sets{}: They are mutable and duplicate elements not allowed(stores unique values)

```python
number ={10, 20, 30,40,50}
print(number)
## set output is unordered like order may vary . No guarantee for order.
```

```python
number ={10, 20, 30,10,40,50} ## Duplicate values are removed automatically
print(number)
```

## Set: Mutable-- Update, add , remove, insert, etc


```python
number ={10, 20, 30,40,50}
value = number.pop()
print(value)
```

```python
number ={10, 20, 30,40,50}
number.pop()
print(number)
```

### Dictionary Method :It stores the data as Key and Value pairs


```python
student ={"name":"Raju",
          "course":"FDE",
          "Age":22
    
}
print(student)
```

```python
## Specific Key calling 
student ={"name":"Raju",
          "course":"FDE",
          "Age":22
    
}
print(student["name"])
```

```python
student ={"name":"Raju",
          "course":"FDE",
          "Age":22
    
}
print(student["Age"])
```
