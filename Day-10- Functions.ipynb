### 1.2 No Parameters with return value:
### This return value can be stored into any variable and we can use into diff functions also.


```python
def veera():
    annual_income= 20000
    monthly_income= annual_income/ 12
    return monthly_income
result =veera()
print(result)
```

```python
def add():
    a=2
    b=5
    c=3
    return(a+b+c)
add()
```

```python
def get_age():
    age = 25 
    return(age)
user_age = get_age()
if user_age>=18:
    print("major")
else:
    print("minor")
```

```python
def calculate_final():
    price =100
    discount = 10
    final_price = price-(price*discount / 100)
    return(final_price)
calculate_final()
```
