## 1.1 String Indexing
### Indexing: we can call individual letter from word 

### Positive index starts from 0 (left).---->
### negative index starts from -1(Right)<-----
## p  Y  T  H  O  N 
## 0  1  2  3  4  5
#-6 -5 -4 -3 -2 -1

```python
# Postive Index
Word = "PYTHON"
print(Word[0])
print(Word[3])
print(Word[5])
```

```python
# Negative Index
Word = "PYTHON"
print(Word[-1])
print(Word[-4])
print(Word[-6])
```

## string slicing
### Slicing means extracting some portion of a string.
### Start -- index where slicing begins, index is included.
### Stop-- index where slicing ends, index is excluded.
### Step --- number of positions to shift.
### Positive Step: start should be less than the end.(left to right) incremental slicing.
### Negative Step: start should be greater than the stop.(right to left) Decremental slicing .
### Ex: [start: stop: step]

```python
word ="python"
print(word[3:4])
```

```python
word ="python"
print(word[3:5])
```

```python
word ="python"
print(word[5:3])
## against the rule so no output
```

```python
word ="python"
print(word[3:8])
## python slicing not give any error if stop range is out of the range and also satrt point larger than end point.
```

```python
word ="python"
print(word[1:5])
```

```python
# Omitting the start index
word ="PYTHON"
print(word[:3])
print(word[:5])
print(word[:8])
```

```python
# Omitting the stop index:
word ="PYTHON"
print(word[3:])
print(word[1:])
print(word[:])
```
