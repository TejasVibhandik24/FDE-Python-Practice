```python
def veera(a, b): # here a, b parameter 
    return a+b   # return keyword give result 
result=veera(10, 20)     # 10, 20 input argument to function call 
print(result)
#note we can use this return value to another function if reuired 
```

```python
def caliculate_bill (price, quantity, discount):  #price 1000 # quantity 3 #discount  10
    total = price * quantity
    discount_amount = total * discount / 100
    final_amount = total - discount_amount
    return final_amount   #i need final_amount as result
bill = caliculate_bill(2, 2000, 20)
print(bill)
```

```python
## example of multiple return values
def calculate_bill(price,quantity,discount):
    total = price*quantity
    discount_amount = total * discount /100
    final_amount = total - discount_amount
    return final_amount,discount  # I need to call multiple values
calculate_bill(100,2,10)
```

```python
def add(a,b):
    return(a-b)
add(b=10,a=20)   
# The order matters here, python assign based on the position of the values.
# Definitely I value will change if you change position of a and b.
# Can overcome by using keyword argument 
```

```python
def student(age, name, course):
    return(age, name,course)
student(course = "FDE",age=20, name = "Test")
```

```python
def add_first_function(a,b):
    return(a+b)
result = add_first_function(3,2)

def mul_second_function(number):
    return number*2
final_result =mul_second_function(result)

def div_third_function(number):
    return number / 3
actual_result=div_third_function(final_result)
print(actual_result)
```
